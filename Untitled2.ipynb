```python

# 1. 라이브러리
import tensorflow as tf
import numpy as np
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

# 2. 데이터 불러오기
iris = load_iris()
X = iris.data
y = iris.target

# 3. 데이터 전처리
scaler = StandardScaler()
X = scaler.fit_transform(X)

# 4. AI 모델 생성
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])

# 5. 모델 학습
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(X, y, epochs=30, verbose=0)

# 6. 모델 정확도
loss, accuracy = model.evaluate(X, y, verbose=0)
print(f"모델 학습 정확도: {accuracy*100:.2f}%")

# 7. 사용자 입력
print("\n===== AI 붓꽃 품종 예측기 =====")
print("꽃의 크기를 cm 단위로 입력하세요.")

sepal_length = float(input("꽃받침 길이: "))
sepal_width = float(input("꽃받침 너비: "))
petal_length = float(input("꽃잎 길이: "))
petal_width = float(input("꽃잎 너비: "))

# 8. AI 예측
user_data = np.array([[
    sepal_length,
    sepal_width,
    petal_length,
    petal_width
]])

user_data = scaler.transform(user_data)

prediction = model.predict(user_data, verbose=0)

result = iris.target_names[np.argmax(prediction)]
confidence = np.max(prediction) * 100

# 9. 결과 출력
print("\n===== 예측 결과 =====")
print("AI가 예측한 품종:", result)
print(f"예측 확률: {confidence:.2f}%")
```
